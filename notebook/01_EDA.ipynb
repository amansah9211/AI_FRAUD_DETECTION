## **Exploratory Data Analysis (EDA)**

Exploratory Data Analysis (EDA) was performed to understand the structure, distribution, and important patterns present in the credit card transaction dataset. The main objective of EDA was to analyze transaction behavior and identify characteristics that could help in detecting fraudulent transactions.


### **1. Dataset Overview**

```python
credit.head(60)
```

```python
credit.columns
```

```python
credit.info()
```

The dataset contains **284,807 credit card transactions** and **31 columns**.

- **Total Records:** 284,807
- **Total Features:** 30
- **Target Variable:** `Class`
- **Data Types:** 30 columns are `float64` and 1 column is `int64`
- **Missing Values:** No missing values were found in the dataset.
- **Memory Usage:** Approximately 67.4 MB

The dataset contains the following features:

- `Time` — Time elapsed between transactions.
- `V1` to `V28` — Anonymized numerical features obtained through Principal Component Analysis (PCA).
- `Amount` — Transaction amount.
- `Class` — Target variable indicating whether a transaction is legitimate or fraudulent.

The `Class` column is the target variable, where:

- `0` → Normal Transaction
- `1` → Fraudulent Transaction

```python
credit.corr(numeric_only=True)['Class'].sort_values(ascending=False)
```

```python
corr_with_class = credit.corr(numeric_only=True)['Class']
top_corr = corr_with_class.abs().sort_values(ascending=False)
print(top_corr)
```

```python
corr = credit.corr(numeric_only=True)['Class'].sort_values()
plt.figure(figsize=(8,10))
corr.plot(kind='barh')
plt.title("Correlation of Features with Class")
plt.xlabel("Correlation")
plt.show()
```

```python
credit.corr(numeric_only=True)['Class'].abs().sort_values(ascending=False).head(10)
```

```python
corr = credit.corr()['Class'].sort_values()
positive_corr = corr[corr > 0]
print(positive_corr)
```

```python
corr = credit.corr()['Class'].sort_values()
negative_corr = corr[corr < 0]
print(negative_corr)
```

```python
credit.groupby('Class').agg(['mean', 'median', 'std'])
```

```python
mean_compare = credit.groupby('Class').mean(numeric_only=True).T
print(mean_compare)
```

```python
mean_values = credit.groupby('Class').mean(numeric_only=True)
difference = (mean_values.loc[1] - mean_values.loc[0]).abs()
difference = difference.sort_values(ascending=False)
print(difference)
```

```python
difference.head(10)
```

```python
difference.head(10).plot(kind='bar')
plt.title("Top 10 Features with Highest Mean Difference")
plt.xlabel("Features")
plt.ylabel("Absolute Mean Difference")
plt.xticks(rotation=45)
plt.show()
```
